### importing the dataset

```python
import numpy as np
import pandas as pd
import torch

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

### train_test_split

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2)
```

Due to irregularity in the values, we need to scale the values for [0, 1]

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

In y_train, there are naming which the neural network can't understand it. So, we need to LabelEncode it.

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

Converting the obtained numpy arrays into pytorch tensors

```python
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)

y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

### Definig the model

```python
class SimpleNN():

    def __init__(self, X):
        self.weights = torch.rand(X.shape[1], 1, dtype=torch.float64, requires_grad=True)
        self.bias = torch.zeros(1, dtype=torch.float64, requires_grad=True)

    def forwardpass(self, X):
        z = torch.matmul(X, self.weights) + self.bias
        y_pred = torch.sigmoid(z)
        return y_pred
    
    # Here we are using BCE as loss function
    def loss_function(self, y_pred, y):
        epsilon = 1e-7
        y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)

        loss = (-1) * ( y_train_tensor * torch.log(y_pred) + (1 - y_train_tensor) * torch.log(1 - y_pred)).mean()
        return loss
```

Imp Parameters

```python
learning_rate = 0.1
epochs = 25
```

Training the pipeline

```python
# creating the model
model = SimpleNN(X_train_tensor)

#define the loop
for epoch in range(epochs):

    # forward pass
    y_pred = model.forwardpass(X_train_tensor)
    
    # loss calculation
    loss = model.loss_function(y_pred, y_train_tensor)
    
    # backward pass
    loss.backward()

    # update the parameters
    with torch.no_grad():
        model.weights -= learning_rate * model.weights.grad
        model.bias -= learning_rate * model.bias.grad

    # zeroing the gradients
    model.weights.grad.zero_()
    model.bias.grad.zero_()

    # print loss in each epoches
    print(f"Epoch : {epoch + 1}, Loss : {loss.item()}")
```

### Model Evaluation

```python
with torch.no_grad():
    # For the test data, we apply the forward pass
    y_pred = model.forwardpass(X_test_tensor)

    # We need to convert the obtain y_pred value of the X_test data, to 0/1 as to compare that with y_test value 
    y_pred = (y_pred > 0.634).float()

    # Accuracy
    accuracy = (y_pred == y_test_tensor).float().mean()
    print(f"Accuracy: {accuracy.item()}")
```
