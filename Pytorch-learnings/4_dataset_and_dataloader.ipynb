In Dataset Class there are 3 major constructs which rule it...
1. __init__ function - initialises the feeatures and labels
2. __len__ function - return the size of the batches
3. getitem function - used to obtain the each element from the batches.

```python
from torch.utils.data import Dataset, DataLoader
```

```python
class CustomDataset(Dataset):

  def __init__(self, features, labels):
    self.features = features
    self.labels = labels

  def __len__(self):
    return self.features.shape[0]

  def __getitem__(self, index):
    return self.features[index], self.labels[index]
```

```python
dataset = CustomDataset(X, y)
```

```python
dataset[2]
```

```python
dataloader = DataLoader(dataset, batch_size = 2, shuffle = True)
```

```python
for batch_features, batch_labels in dataloader:
  print(batch_features)
  print(batch_labels)
  print("-"*50)
```
