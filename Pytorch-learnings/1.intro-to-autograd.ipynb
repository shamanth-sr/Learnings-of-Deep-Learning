### Simple Single Perceptron

Lets test a Autograd on a single Perceptron structure with,
x = 6.7
y = 0

w = 1
b = 0

The sequence of calculation is obtaining z = w*x + b, y_pred = sigmoid(z), loss = binary_cross_entropy(y_pred, y)

```python
x = torch.tensor(6.7)
y = torch.tensor(0)
```

```python
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

print("w :", w)
print("b :", b)
```

```python
z = w * x + b
print("z = ", z)
```

```python
y_pred = torch.sigmoid(z)
print("y_pred = ", y_pred)
```

```python
# defining binary_cross_entropy function

def binary_cross_entropy_loss(prediction, target):
    epsilon = 1e-8
    prediction = torch.clamp(prediction, epsilon, 1-epsilon)
    return (-1)*(target * torch.log(prediction) + (1 - target) * torch.log(1 - prediction))
```

```python
# importing binary-cross-entropy-loss function
import torch.nn as nn

lossfn = nn.BCELoss()

loss = binary_cross_entropy_loss(y_pred, y)

loss
```

```python
loss.backward()
```

```python
print("Gradient of w = ", w.grad)
print("Gradient of b = ", b.grad)
```

So, from this we got the value of z, y_pred and loss too. When 
x = 6.7

### Clearing Gradien Value

```python
x = torch.tensor(2.0, requires_grad=True)
print("x = ", x)
```

```python
y = x ** 2
print("y = ", y)
```

```python
y.backward()
```

```python
x.grad
```

On repeating running the last three blocks of code, we get altering values of 'x', beacuse the gradient value of x are still stored in that, and we again run the code it get add to the previous value.

So, we introduce grad.zero_()

```python
x.grad.zero_()
```

### How we can stop the gradient

1. using require_grad = False
2. detach()
3. torch.no_grad()

```python
x = torch.tensor(2.0, requires_grad=True)
x
```

```python
z = x.detach()
```

```python
y = x ** 2
y
```

```python
y1 = z ** 2
y1
```

```python
y.backward()
```

```python
y1.backward()
```

So, here y1 = z ** 2, and z is detached from backward function
