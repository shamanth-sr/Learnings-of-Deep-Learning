In this learnig we are going to learn about the nn module of Pytorch, by creating a single perceptron architecture which is used for bianry classification problem.

```python
# creating model class

import torch
import torch.nn as nn

class Model(nn.Module):

  # Defining the class imported from nn module
  def __init__(self, num_features):
    super().__init__()
    self.Linear = nn.Linear(num_features, 1)
    self.sigmoid = nn.Sigmoid()

  # Creating a forward pass function
  def forward(self, features):
    out = self.Linear(features)
    out = self.sigmoid(out)
    return out
```

```python
# create a random dataset
features = torch.rand(10, 5)

# create model
model = Model(features.shape[1])

#calling the forward pass
''' model.forward(features) can also be used to trigger the forward-pass
function, and there is a internal function called __call__ which will automatically triggers the forward pass itself '''
model.forward(features)

```

```python
# show model weights
model.Linear.weight
```

```python
#show model bias
model.Linear.bias
```

Now, we make some tiny changes in the network by adding a hidden layer and also increasing the input-features

```python
class Model(nn.Module):

  def __init__(self, num_features):
    super().__init__()
    self.Linear1 = nn.Linear(num_features, 3)
    self.relu = nn.ReLU()

    self.Linear2 = nn.Linear(3, 1)
    self.sigmoid = nn.Sigmoid()

  def forward(self, features):
    out = self.Linear1(features)
    out = self.relu(out)

    out = self.Linear2(out)
    out = self.sigmoid(out)

    return out
```

```python
model = Model(features.shape[1])

model(features)
```

```python
summary(model, input_size=(10, 5))
```

Sequential Pipeline.....

```python
class Model(nn.Module):

  def __init__(self, num_features):
    super().__init__()
    self.network = nn.Sequential(
        nn.Linear(num_features, 3),
        nn.ReLU(),
        nn.Linear(3, 1),
        nn.Sigmoid()
    )

  def forward(self, features):
    out = self.network(features)

    return out
```

```python
model = Model(features.shape[1])

model(features)
```
